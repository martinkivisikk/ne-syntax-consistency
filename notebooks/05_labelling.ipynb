# NE näidete märgendamine

Käib läbi `scripts/export_examples.py` tehtud näited ükshaaval: lause sõltuvusjoonis
(span esile tõstetud), lause `[[...]]` märgistusega ja väljad `ne`, `ne_fix`, `note`.
Reeglid: `labelling_guidelines.md`.

- **save ▶ next** salvestab rea kohe CSV-sse ja läheb edasi; vigane märgend jääb
  salvestamata ja põhjus kuvatakse punaselt.
- **◀ prev**, **next unlabelled** ja **row … go** salvestavad samuti enne liikumist,
  kui `ne` on valitud.
- **CHECK** lisab märkusele eesliite `CHECK: ` (piirjuhud, mida arutada).
- `boundary` valimisel pakutakse `ne_fix`-i algväärtuseks span ilma ääre-kirjavahemärkideta.
- Vihiku võib igal hetkel sulgeda; uuesti käivitades alustab esimesest märgendamata reast.


```python
from pathlib import Path

from IPython.display import display

from ne_syntax_consistency.labelling import Labeller, load_labels, load_sentences


def _project_root() -> Path:
    # Jupyter käivitab kerneli notebooki kaustas; otsime üles pyproject.toml-i
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p / "pyproject.toml").exists():
            return p
    raise FileNotFoundError("pyproject.toml not found")


ROOT = _project_root()
```

```python
# Parameetrid
FILE = "results/span_trees/examples_type_full_nochat.csv"  # märgendatav fail
SENTENCES = "data/full"  # lausete Parquet-fail või kaust (joonise jaoks)
WINDOW = 5  # sõnu iga spanni sõna ümber; "whole sentence" näitab tervet lauset
```

```python
labels = load_labels(ROOT / FILE)
sentences = load_sentences(ROOT / SENTENCES, labels["text_id"])
print(
    f"{(labels['ne'] != '').sum()} / {len(labels)} märgendatud, {len(sentences)} lauset"
)
```

```python
labeller = Labeller(labels, ROOT / FILE, sentences, WINDOW)
display(labeller.widget)
```
